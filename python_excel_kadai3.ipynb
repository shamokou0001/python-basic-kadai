```python
import pandas as pd
import openpyxl
from openpyxl.styles import PatternFill
```

```python
# ファイルリスト
# ファイルがGoogle Driveのルートにあると仮定しています。もし別のフォルダにある場合は、パスを適宜変更してください。
file_list = ["/content/drive/MyDrive/2022_年間売上表.xlsx", "/content/drive/MyDrive/2023_年間売上表.xlsx"]

# データフレームを格納するための空のデータフレーム
df = pd.DataFrame()

for file_name in file_list:
    # Excelファイルを読み込み（ヘッダーあり）
    temp_df = pd.read_excel(file_name)

    # 列名チェックと修正
    temp_df.columns = ['商品', '売上年', '金額（千円）']

    # 金額列を数値型に変換
    temp_df["金額（千円）"] = pd.to_numeric(temp_df["金額（千円）"], errors="coerce")

    # データフレームを結合
    df = pd.concat([df, temp_df], ignore_index=True)
```

```python
# 集計
result = df.groupby(["商品", "売上年"], as_index=False).agg({"金額（千円）": "sum"})
```

```python
output_file = "/content/drive/MyDrive/売上集計表.xlsx"
with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    result.to_excel(writer, sheet_name="売上集計", index=False)

# 既存のExcelファイルを開いてヘッダーに色をつける
workbook = openpyxl.load_workbook(output_file)
sheet = workbook['売上集計']

for cell in sheet["1:1"]:
    cell.fill = PatternFill(patternType='solid', fgColor='F2F2F2')

workbook.save(output_file)
```
