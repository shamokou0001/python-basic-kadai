```python
from openpyxl import Workbook
from datetime import datetime

# Excelブック作成
wb = Workbook()
ws = wb.active

# タイトル
ws["B2"] = "請求書"

# 会社情報
ws["B4"] = "株式会社ABC"
ws["B5"] = "〒101-0022 東京都千代田区神田練塀町300"
ws["B6"] = "TEL: 03-1234-5678  FAX: 03-1234-5678"
ws["B7"] = "担当者名: 鈴木一郎 様"

# 右側情報
ws["F4"] = "No."
ws["G4"] = "0001"
ws["F5"] = "日付"
ws["G5"] = datetime.today().strftime("%Y/%m/%d")

# 表ヘッダー
ws["B10"] = "商品名"
ws["C10"] = "数量"
ws["D10"] = "単価"
ws["E10"] = "金額"

# 商品データ
ws["B11"] = "商品A"
ws["C11"] = 2
ws["D11"] = 10000
ws["E11"] = 20000

ws["B12"] = "商品B"
ws["C12"] = 1
ws["D12"] = 15000
ws["E12"] = 15000

# 合計計算
ws["E13"] = "=SUM(E11:E12)"

# 合計
ws["B15"] = "合計"
ws["E15"] = "=E13"

# 消費税
ws["B16"] = "消費税"
ws["E16"] = "=E15*0.1"

# 税込合計
ws["B17"] = "税込合計"
ws["E17"] = "=E15+E16"

# ファイル名（現在日付）
filename = "請求書_" + datetime.today().strftime("%Y%m%d") + ".xlsx"

# 保存
wb.save(filename)

print(filename + " を作成しました")
```
